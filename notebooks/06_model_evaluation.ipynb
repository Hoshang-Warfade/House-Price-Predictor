## Evaluation Metrics

To evaluate the performance of the regression model, the following metrics are used. Each metric provides a different perspective on the model's predictive performance, so they should be interpreted together rather than individually.

---

### 1. Mean Absolute Error (MAE)

#### Definition

Mean Absolute Error (MAE) measures the average absolute difference between the actual and predicted target values.

It considers only the magnitude of prediction errors, treating overestimation and underestimation equally.

A lower MAE indicates better model performance.

#### Formula

$$
MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

#### Advantages

- Easy to interpret.
- Expressed in the same unit as the target variable.
- Less sensitive to outliers than MSE and RMSE.

#### Limitation

MAE does not place additional emphasis on large prediction errors. As a result, it may not be suitable for applications where large errors are significantly more costly than small ones.

---

### 2. Mean Squared Error (MSE)

#### Definition

Mean Squared Error (MSE) measures the average squared difference between the actual and predicted target values.

By squaring prediction errors before averaging, MSE penalizes large errors more heavily than small ones, making it more sensitive to outliers.

A lower MSE indicates better model performance.

#### Formula

$$
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

#### Advantages

- Penalizes large prediction errors more heavily.
- Differentiable, making it suitable as an optimization objective for many machine learning algorithms.

#### Limitation

Since prediction errors are squared, large errors receive disproportionately higher penalties. As a result, MSE is highly sensitive to outliers.

---

### 3. Root Mean Squared Error (RMSE)

#### Definition

Root Mean Squared Error (RMSE) is the square root of the Mean Squared Error (MSE).

It expresses the prediction error in the same unit as the target variable while still penalizing large prediction errors more heavily.

A lower RMSE indicates better model performance.

#### Formula

$$
RMSE=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}
$$

#### Advantages

- Expressed in the same unit as the target variable.
- Easier to interpret than MSE.
- Penalizes large prediction errors more heavily than MAE.

#### Limitation

Since RMSE is based on squared errors, it remains sensitive to outliers. Large prediction errors have a greater influence on the metric than small errors.

---

### 4. R² Score (Coefficient of Determination)

#### Definition

The R² score measures the proportion of the variance in the target variable that is explained by the regression model.

It compares the model's predictions against a baseline model that always predicts the mean of the target variable.

An R² score closer to **1** indicates that the model explains a larger proportion of the variability in the target.

#### Formula

$$
R^2 = 1-\frac{SS_{res}}{SS_{tot}}
$$

where,

$$
SS_{res}=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

$$
SS_{tot}=\sum_{i=1}^{n}(y_i-\bar{y})^2
$$

#### Advantages

- Unitless metric.
- Easy to compare different regression models.
- Indicates how much variance in the target is explained by the model.

#### Limitation

R² generally increases or remains unchanged when additional features are added, even if they do not improve the model's predictive ability. Therefore, it may overestimate the performance of more complex models.

---

### 5. Adjusted R² Score

#### Definition

Adjusted R² is a modified version of the R² score that accounts for the number of predictor variables in the model.

Unlike R², Adjusted R² increases only when newly added features provide a meaningful improvement in the model's performance. It may decrease if additional features do not contribute useful information.

As a result, Adjusted R² is a more reliable metric for comparing regression models with different numbers of predictors.

#### Formula

$$
Adjusted\ R^2=
1-(1-R^2)\times\frac{n-1}{n-p-1}
$$

where,

- **n** = Number of observations
- **p** = Number of predictor variables

#### Advantages

- Penalizes unnecessary predictor variables.
- Better suited for comparing regression models with different numbers of features.
- Encourages simpler models without sacrificing predictive performance.

#### Limitation



Adjusted R² accounts for model complexity by penalizing the inclusion of additional predictors during model fitting. Since this adjustment is based on the data used to estimate the model parameters, it does not directly measure how well the model generalizes to unseen data. Therefore, Adjusted R² is primarily reported for the training data, while model performance on validation or test datasets is better assessed using prediction-based metrics such as MAE, RMSE, and R².

### Step 06.2: Evaluate the Baseline Model Using the Reusable Evaluation Function

To ensure consistency and avoid code duplication, a reusable evaluation function is used to compute the regression metrics for the baseline model.

The function calculates the following metrics:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R² Score
- Adjusted R² Score

The same function is applied to both the training and validation predictions, ensuring a consistent evaluation process across different regression models. This modular approach improves code reusability, maintainability, and simplifies future model comparisons.

Since the model is trained on the transformed feature space, the number of predictor variables is obtained from the fitted preprocessing pipeline and used to calculate the Adjusted R² score.


````python
#Import evaluation fun
from src.evaluation import evaluate_regression_model
pd.options.display.float_format = "{:.4f}".format

# Number of features after preprocessing
n_features = len(
    pipeline.named_steps["column_transformer"]
    .get_feature_names_out()
)

# Training metrics
train_metrics = evaluate_regression_model(
    y_true=y_train,
    y_pred=y_train_pred,
    n_features=n_features,
    adjusted_r2=True,
)
# Validation metrics
validation_metrics = evaluate_regression_model(
    y_true=y_valid,
    y_pred=y_valid_pred,
)
````

### Step 06.3: Summarize & Interprete Evaluation Results

The evaluation metrics for the training and validation datasets are consolidated into a single summary table.

Presenting the metrics in a tabular format enables a clear comparison of model performance across datasets and provides a baseline for evaluating and comparing future regression models.

````python
baseline_model_evaluation_summary = pd.DataFrame(
    {
        "Training": train_metrics,
        "Validation": validation_metrics,
    }
).round(4)

baseline_model_evaluation_summary
````

### Interpretation

The baseline Linear Regression model achieves strong performance on the training dataset, with an **R² score of 0.9401**, indicating that it explains approximately **94.01%** of the variance in the training data. The relatively low MAE and RMSE further suggest that the model fits the training data well.

However, the model's performance deteriorates on the validation dataset. The validation **R² score decreases to 0.4430**, while both the MAE and RMSE increase substantially. This indicates that the model is less effective at predicting unseen data.

The noticeable gap between the training and validation metrics suggests that the baseline model is experiencing **overfitting (high variance)**. Although it captures the patterns present in the training data effectively, it does not generalize well to new observations.

Overall, the Linear Regression model provides a useful baseline for comparison with more advanced regression models. In the subsequent steps, regularized regression models such as Ridge, Lasso, and ElasticNet will be evaluated to improve generalization performance and reduce overfitting.
